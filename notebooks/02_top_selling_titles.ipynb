# 02 – Nintendo Top-Selling Games

## Project Goal

This notebook explores the five best-selling games for Nintendo Switch 2, Nintendo Switch and Wii U.

The goal is to compare the sales figures of the most successful games across these platforms.

## Data Source

Source: [Nintendo – Dedicated Video Game Sales Units](https://www.nintendo.co.jp/ir/en/finance/software/index.html)

Data as of: September 30, 2026 (Switch 2 and Switch) and March 31, 2026 (Wii U).

The dataset contains the five best-selling games for each selected platform.

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt

top_games = df[df["rank"] == 1]

plt.figure(figsize=(9, 5))

bars = plt.bar(
    top_games["platform"],
    top_games["units_million"]
)

plt.title("Best-Selling Game by Nintendo Platform")
plt.xlabel("Nintendo Platform")
plt.ylabel("Units Sold (Millions)")

plt.bar_label(bars, fmt="%.2f", padding=3)

plt.tight_layout()
plt.savefig("../reports/figures/top_selling_games.png", dpi=300)
plt.show()
```

## Interpretation

The Nintendo Switch has the highest combined sales among the five best-selling games, with 226.47 million units.

The Switch 2 follows with 32.03 million units, while the Wii U reaches 30.76 million units.

Mario Kart is the best-selling game on all three platforms included in this analysis.

However, these figures represent only the five best-selling games per platform and should not be interpreted as total software sales.

Note: Nintendo's figures include games bundled with hardware. Mario Kart World was sold in a bundle with the Switch 2, which increases its sales figure.
