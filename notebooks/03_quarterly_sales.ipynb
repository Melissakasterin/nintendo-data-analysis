# 03 – Nintendo Switch 2: Quarterly Hardware Sales

This notebook shows how many Nintendo Switch 2 consoles were sold in each quarter since launch.

**Source:** Nintendo earnings releases and presentations ([Q1 FY27 presentation](https://www.nintendo.co.jp/ir/pdf/2026/260806_2e.pdf), [H1 FY27 earnings release](https://www.nintendo.co.jp/ir/pdf/2026/261105e.pdf))

**Note:** Nintendo's fiscal year starts in April. FY27/Q2 was calculated as the six-month figure (6.49) minus FY27/Q1 (3.82).

```python
import pandas as pd

df = pd.read_csv("../data/raw/switch2_quarterly_hardware.csv")

df
```

```python
df["hardware_million"].sum()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(df["months"], df["hardware_million"], marker="o")

plt.title("Nintendo Switch 2: Hardware Sales per Quarter")
plt.xlabel("Quarter")
plt.ylabel("Units Sold (Millions)")

plt.tight_layout()
plt.savefig("../reports/figures/switch2_quarterly_hardware.png", dpi=300)
plt.show()
```

## Interpretation

- The highest sales were in Oct–Dec 2025 with 7.01 million units. This is the holiday quarter.
- The lowest sales were in Jan–Mar 2026 with 2.49 million units, right after the holiday season.
- In the most recent quarter (Jul–Sep 2026), sales were 2.67 million units, compared to 4.54 million in the same quarter one year earlier.

**Hypotheses (not proven by this data):** possible reasons for the decline include rising component costs and the end of the launch period. The data shows *what* happened, not *why*.
